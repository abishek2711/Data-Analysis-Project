```python
import pandas as pd
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv(r"C:\Users\HP\Downloads\india_job_market_dataset.csv")
```

DATA CLEANSING AND PROCESSING

```python
# Convert dates to datetime format
data["Posted Date"] = pd.to_datetime(data["Posted Date"])
data["Application Deadline"] = pd.to_datetime(data["Application Deadline"])
```

DATA VISUALIZATION AND ANALYSIS

```python
# Top 10 Job Titles
top_jobs = data["Job Title"].value_counts().head(10)
plt.figure(figsize=(12, 5))
sns.barplot(x=top_jobs.values, y=top_jobs.index, hue=None, palette="viridis",  legend=False)
plt.xlabel("Number of Job Listings")
plt.ylabel("Job Title")
plt.title("Top 10 Most Common Job Titles in India")
plt.show()
```

```python
# Job Listings by Location
top_locations = data["Job Location"].value_counts().head(10)
plt.figure(figsize=(10, 5))
sns.barplot(x=top_locations.values, y=top_locations.index, palette="coolwarm")
plt.xlabel("Number of Jobs")
plt.ylabel("Location")
plt.title("Top 10 Job Locations in India")
plt.show()
```

```python
# Job Listings by Experience Level
exp_level = data["Experience Required"].value_counts()
plt.figure(figsize=(8, 5))
sns.barplot(x=exp_level.index, y=exp_level.values, palette="magma")
plt.xlabel("Experience Level")
plt.ylabel("Number of Jobs")
plt.title("Job Listings by Experience Level")
plt.xticks(rotation=45)
plt.show()
```

```python
# Job Distribution by Job Portal
portal_counts = data["Job Portal"].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(portal_counts, labels=portal_counts.index, autopct="%1.1f%%", colors=["blue", "green", "red"], startangle=90)
plt.title("Job Listings Distribution by Job Portal")
plt.show()
```

```python
# Remote vs Onsite Jobs
remote_jobs = data["Remote/Onsite"].value_counts()
plt.figure(figsize=(7, 5))
sns.barplot(x=remote_jobs.index, y=remote_jobs.values, palette="Set2")
plt.xlabel("Job Type")
plt.ylabel("Number of Jobs")
plt.title("Remote vs Onsite Job Listings")
plt.show()
```
